```python
# prompt: create dictionaly with one = 30, two=20 and three= 50 and plot it as a pie chart

import matplotlib.pyplot as plt

# Create the dictionary
data = {'one': 30, 'two': 20, 'three': 50}

# Extract labels and values for the pie chart
labels = list(data.keys())
values = list(data.values())

# Create the pie chart
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90, explode=(0, 0.2, 0), shadow = True)

# Add a title
plt.title('Pie Chart of Data')

# Display the chart
plt.show()
```
